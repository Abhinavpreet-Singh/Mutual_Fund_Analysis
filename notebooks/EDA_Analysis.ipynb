# Day 3 Exploratory Data Analysis

Deep EDA across NAV, AUM, SIP, investor, portfolio, and performance datasets. The notebook generates 15+ charts and saves publication-quality exports to `reports/day3_charts/`.

```python
import sys
from pathlib import Path
from IPython.display import display, Image, HTML

NOTEBOOK_ROOT = Path.cwd()
if not (NOTEBOOK_ROOT / 'day3_eda.py').exists():
    for candidate in [NOTEBOOK_ROOT, *NOTEBOOK_ROOT.parents]:
        if (candidate / 'day3_eda.py').exists():
            NOTEBOOK_ROOT = candidate
            break
sys.path.insert(0, str(NOTEBOOK_ROOT))

import day3_eda as e

data = e.load_datasets()

def show_artifacts(paths):
    for path in paths:
        artifact = Path(path)
        print(artifact.name)
        if artifact.suffix.lower() == '.png':
            display(Image(filename=str(artifact)))
        elif artifact.suffix.lower() == '.html':
            display(HTML(artifact.read_text(encoding='utf-8')))
        else:
            print(artifact.read_text(encoding='utf-8'))

print(f'Project root: {e.BASE_DIR}')
for name, frame in data.items():
    print(f'{name:<15} {frame.shape[0]:>8,} rows x {frame.shape[1]} cols')
```

## 1. NAV Trend Analysis

All 40 schemes are plotted across the 2022-2026 window, with the 2022 recovery base, 2023 rally, and 2024 correction bands highlighted.

```python
nav_paths = e.nav_charts(data)
show_artifacts(nav_paths)
```

## 2. AUM and SIP Trends

Grouped AUM bars, monthly SIP inflow time series, and category inflow heatmap.

```python
aum_sip_paths = e.aum_sip_charts(data)
show_artifacts(aum_sip_paths)
```

## 3. Investor Demographics and Geography

Age distribution, SIP amount by age, geographic concentration, and T30 vs B30 split.

```python
demo_paths = e.demographic_geo_charts(data)
show_artifacts(demo_paths)
```

## 4. Portfolio and Correlation

Folio growth, NAV-return correlation structure, and equity sector allocation across portfolios.

```python
portfolio_paths = e.portfolio_charts(data)
show_artifacts(portfolio_paths)
```

## 5. Performance and Risk

Return/risk scatterplots and the risk grade distribution summarize the behavior of the scheme universe.

```python
performance_paths = e.performance_charts(data)
show_artifacts(performance_paths)
```

## EDA Findings

1. All 40 schemes are represented in the NAV dataset, and the cleaned series now extends from 2022-01 through 2026-05.
2. SBI Mutual Fund remains the largest AMC in the sample; its latest annual AUM snapshot is about ₹1,250,000 crore.
3. Monthly SIP inflow is structurally higher by the end of the sample, with the December 2025 milestone at roughly ₹31,002 crore.
4. The industry folio base roughly doubled from 13.26 crore to 26.12 crore over the period.
5. The strongest equity portfolio sector exposure is concentrated in Banking, showing that sector allocation is still clustered.
6. SIP value is led by Madhya Pradesh, indicating that transaction value is concentrated in a handful of large investor states.
7. T30 investors still contribute the larger SIP pool versus B30 in absolute amount terms, even though B30 activity is broad-based.
8. The highest Sharpe ratio in the dataset belongs to ICICI Pru Liquid Fund - Regular - Growth (7.68), which is consistent with liquid-fund style stability.
9. The lowest expense ratio fund in the sample is Nippon India Gilt Securities Fund - Regular - Growth at 0.55%, confirming the direct-plan cost advantage.
10. Return and risk remain coupled: SBI Small Cap Fund - Regular Plan - Growth leads on 3-year return at 23.39%, while HDFC Top 100 Fund - Direct Plan - Growth has the shallowest drawdown at -33.50%.
